```python
sales_clean["date"] = pd.to_datetime(
    sales_clean["date"],
    dayfirst=True,
    errors="coerce"
)
```

```python
print("Invalid dates:", sales_clean["date"].isna().sum())
print("Start:", sales_clean["date"].min())
print("End:", sales_clean["date"].max())
```

```python
numeric_columns = [
    "year",
    "age",
    "quantity",
    "price_per_unit",
    "total_amount"
]

for col in numeric_columns:
    sales_clean[col] = pd.to_numeric(
        sales_clean[col],
        errors="coerce"
    )
```

```python
print("Quantity <= 0:", (sales_clean["quantity"] <= 0).sum())
print("Price <= 0:", (sales_clean["price_per_unit"] <= 0).sum())
print("Total <= 0:", (sales_clean["total_amount"] <= 0).sum())
print("Age < 18:", (sales_clean["age"] < 18).sum())
print("Age > 100:", (sales_clean["age"] > 100).sum())
```

```python
sales_clean["calculated_total"] = (
    sales_clean["quantity"] *
    sales_clean["price_per_unit"]
)

sales_clean["amount_difference"] = (
    sales_clean["total_amount"] -
    sales_clean["calculated_total"]
)

print(
    "Mismatched transactions:",
    (sales_clean["amount_difference"].abs() > 0.01).sum()
)
```

```python
sales_clean["month"] = sales_clean["date"].dt.month
sales_clean["year"] = sales_clean["date"].dt.year

sales_clean["year_month"] = (
    sales_clean["date"]
    .dt.to_period("M")
    .astype(str)
)
```

```python
monthly_demand = (
    sales_clean
    .groupby(
        ["year_month", "product_category"],
        as_index=False
    )
    .agg(
        units_sold=("quantity", "sum"),
        total_sales=("total_amount", "sum"),
        transaction_count=("transaction_id", "nunique"),
        average_price=("price_per_unit", "mean"),
        average_age=("age", "mean")
    )
)
```

```python
monthly_demand = monthly_demand.sort_values(
    ["product_category", "year_month"]
)

monthly_demand["demand_growth"] = (
    monthly_demand
    .groupby("product_category")["units_sold"]
    .pct_change() * 100
)
```

```python
monthly_demand["demand_moving_avg_3m"] = (
    monthly_demand
    .groupby("product_category")["units_sold"]
    .transform(
        lambda x: x.rolling(3, min_periods=1).mean()
    )
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

amazon_df = marketplace_products[
    marketplace_products["source"] == "Amazon"
].copy()

flipkart_df = marketplace_products[
    marketplace_products["source"] == "Flipkart"
].copy()

# TF-IDF on product names
vectorizer = TfidfVectorizer(
    ngram_range=(1, 2),
    min_df=1
)

combined_names = pd.concat([
    amazon_df["normalized_name"],
    flipkart_df["normalized_name"]
])

tfidf_matrix = vectorizer.fit_transform(combined_names)

amazon_vectors = tfidf_matrix[:len(amazon_df)]
flipkart_vectors = tfidf_matrix[len(amazon_df):]

# Similarity between every Amazon and Flipkart product
similarity_matrix = cosine_similarity(
    flipkart_vectors,
    amazon_vectors
)

# Best Amazon match for every Flipkart product
best_match_index = similarity_matrix.argmax(axis=1)
best_scores = similarity_matrix.max(axis=1)

flipkart_df["matched_amazon_index"] = best_match_index
flipkart_df["similarity_score"] = best_scores

flipkart_df["matched_amazon_product"] = [
    amazon_df.iloc[i]["product_name"]
    for i in best_match_index
]

display(
    flipkart_df[
        [
            "product_name",
            "matched_amazon_product",
            "similarity_score",
            "category"
        ]
    ]
    .sort_values("similarity_score", ascending=False)
    .head(30)
)
```

```python
import re

def extract_brand(name):
    name = str(name).strip()
    return name.split()[0].lower() if name else ""


def extract_model(name):
    name = str(name).lower()

    # Look for model-like identifiers containing letters + numbers
    models = re.findall(
        r'\b[a-z]{1,6}[-]?[a-z]?\d+[a-z0-9-]*\b',
        name
    )

    return " ".join(models[:5])


marketplace_products["brand_candidate"] = (
    marketplace_products["product_name"]
    .apply(extract_brand)
)

marketplace_products["model_candidate"] = (
    marketplace_products["product_name"]
    .apply(extract_model)
)

display(
    marketplace_products[
        [
            "source",
            "product_name",
            "brand_candidate",
            "model_candidate"
        ]
    ].head(30)
)
```

```python
def calculate_match_score(name_similarity, brand_match, model_match, category_match):

    score = (
        0.50 * name_similarity +
        0.20 * brand_match +
        0.20 * model_match +
        0.10 * category_match
    )

    return score
```

```python
# Prepare Amazon lookup
amazon_lookup = amazon_df.reset_index(drop=True)

results = []

for _, row in flipkart_df.iterrows():

    amazon_row = amazon_lookup.iloc[
        int(row["matched_amazon_index"])
    ]

    # Brand comparison
    flip_brand = extract_brand(row["product_name"])
    amazon_brand = extract_brand(amazon_row["product_name"])

    brand_match = int(
        flip_brand == amazon_brand
        and flip_brand != ""
    )

    # Model comparison
    flip_model = extract_model(row["product_name"])
    amazon_model = extract_model(amazon_row["product_name"])

    if flip_model and amazon_model:
        model_match = int(
            flip_model == amazon_model
        )
    else:
        # If model isn't available in either product,
        # don't give a positive model-match signal
        model_match = 0

    # Category comparison
    category_match = int(
        row["category"] == amazon_row["category"]
    )

    final_score = calculate_match_score(
        row["similarity_score"],
        brand_match,
        model_match,
        category_match
    )

    results.append({
        "flipkart_product": row["product_name"],
        "amazon_product": amazon_row["product_name"],
        "name_similarity": row["similarity_score"],
        "brand_match": brand_match,
        "model_match": model_match,
        "category_match": category_match,
        "final_match_score": final_score
    })

match_results = pd.DataFrame(results)

display(
    match_results
    .sort_values("final_match_score", ascending=False)
    .head(30)
)
```

```python
def match_decision(score):
    if score >= 0.85:
        return "HIGH_CONFIDENCE"

    elif score >= 0.65:
        return "REVIEW"

    else:
        return "NO_MATCH"


match_results["match_decision"] = (
    match_results["final_match_score"]
    .apply(match_decision)
)

print(
    match_results["match_decision"]
    .value_counts()
)
```

```python
def calculate_category_novelty(group):

    texts = group["feature_text"].tolist()

    if len(texts) == 1:
        group["novelty_score"] = 100.0
        return group

    vectorizer = TfidfVectorizer(
        ngram_range=(1, 2),
        min_df=1
    )

    vectors = vectorizer.fit_transform(texts)

    similarity = cosine_similarity(vectors)

    novelty_scores = []

    for i in range(len(texts)):

        # Remove similarity with itself
        other_similarities = np.delete(
            similarity[i],
            i
        )

        average_similarity = (
            other_similarities.mean()
        )

        novelty = (
            1 - average_similarity
        ) * 100

        novelty_scores.append(novelty)

    group["novelty_score"] = novelty_scores

    return group
```

```python
import re

def clean_feature_text(text):
    text = str(text).lower()

    # Remove common brand/model noise
    text = re.sub(r'\b\d+([a-z]+)?\b', ' ', text)

    # Remove very common generic words
    stop_words = [
        "new", "latest", "original", "black", "white",
        "blue", "red", "green", "pack", "pcs",
        "inch", "inches", "cm", "mm"
    ]

    for word in stop_words:
        text = re.sub(
            r'\b' + re.escape(word) + r'\b',
            ' ',
            text
        )

    text = re.sub(r'[^a-z\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()

    return text


product_master["novelty_text"] = (
    product_master["product_name"]
    .fillna("")
    .apply(clean_feature_text)
)

display(
    product_master[
        ["product_name", "novelty_text"]
    ].head(20)
)
```

```python
def calculate_customer_value(
    price,
    current_solution_cost,
    time_saved_hours,
    hourly_value_of_time
):

    cost_saved = max(
        current_solution_cost - price,
        0
    )

    economic_value_of_time = (
        time_saved_hours * hourly_value_of_time
    )

    customer_value = (
        cost_saved +
        economic_value_of_time
    )

    value_to_price_ratio = (
        customer_value / price
        if price > 0
        else 0
    )

    return {
        "cost_saved": round(cost_saved, 2),
        "economic_value_of_time": round(
            economic_value_of_time, 2
        ),
        "customer_value": round(
            customer_value, 2
        ),
        "value_to_price_ratio": round(
            value_to_price_ratio, 3
        )
    }
```

```python
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

COMPETITOR_THRESHOLD = 0.50

# Reset competition columns
product_master["competitor_count"] = 0
product_master["competition_similarity"] = 0.0
product_master["competitor_avg_price"] = np.nan

# Calculate competition category-wise
for category, group in product_master.groupby("category"):

    indices = group.index.tolist()

    if len(indices) < 2:
        continue

    texts = group["competition_text"].fillna("").tolist()

    vectorizer = TfidfVectorizer(
        stop_words="english",
        ngram_range=(1, 2),
        min_df=1
    )

    tfidf = vectorizer.fit_transform(texts)

    similarity_matrix = cosine_similarity(tfidf)

    prices = group["price"].to_numpy()

    for i, idx in enumerate(indices):

        # Remove the product itself
        other_positions = [j for j in range(len(indices)) if j != i]

        similarities = similarity_matrix[i, other_positions]

        # Select only competitors >= 0.50
        competitor_mask = similarities >= COMPETITOR_THRESHOLD

        competitor_count = int(competitor_mask.sum())

        product_master.loc[idx, "competitor_count"] = competitor_count

        if competitor_count > 0:

            # Similarity ONLY of selected competitors
            selected_similarities = similarities[competitor_mask]

            product_master.loc[idx, "competition_similarity"] = round(
                selected_similarities.mean(), 4
            )

            # Get positions of selected competitors
            selected_positions = np.array(other_positions)[competitor_mask]

            competitor_prices = prices[selected_positions]

            # Remove missing prices
            competitor_prices = competitor_prices[
                ~pd.isna(competitor_prices)
            ]

            if len(competitor_prices) > 0:

                product_master.loc[idx, "competitor_avg_price"] = round(
                    competitor_prices.mean(), 2
                )
```

```python
max_competitors = product_master["competitor_count"].max()

if max_competitors > 0:
    product_master["competition_intensity"] = (
        product_master["competitor_count"]
        / max_competitors
        * 100
    ).round(2)
else:
    product_master["competition_intensity"] = 0
```

```python
# Initialize substitute columns
product_master["substitute_count"] = 0
product_master["substitute_avg_price"] = np.nan
product_master["substitute_pressure"] = 0.0

# Reverse mapping:
# product_type -> substitute product types
product_to_substitutes = {}

for group_name, product_types in substitute_groups.items():
    for product_type in product_types:
        product_to_substitutes[product_type] = [
            x for x in product_types
            if x != product_type
        ]

# Calculate substitutes
for idx, row in product_master.iterrows():

    product_type = row["product_type"]

    # Skip products without a defined product type
    if not product_type or product_type not in product_to_substitutes:
        continue

    substitute_types = product_to_substitutes[product_type]

    # Find products belonging to substitute types
    substitutes = product_master[
        product_master["product_type"].isin(substitute_types)
    ]

    if len(substitutes) == 0:
        continue

    product_master.loc[idx, "substitute_count"] = len(substitutes)

    # Average price of substitutes
    substitute_prices = substitutes["price"].dropna()

    if len(substitute_prices) > 0:
        product_master.loc[idx, "substitute_avg_price"] = round(
            substitute_prices.mean(), 2
        )
```

```python
def get_google_trends(keyword):

    pytrends.build_payload(
        [keyword],
        cat=0,
        timeframe="today 5-y",
        geo="IN",
        gprop=""
    )

    data = pytrends.interest_over_time()

    if data.empty:
        return pd.DataFrame()

    data = data.reset_index()

    if "isPartial" in data.columns:
        data = data.drop(columns=["isPartial"])

    data = data.rename(
        columns={
            "date": "date",
            keyword: "search_interest"
        }
    )

    return data
```

```python
def get_indian_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Summer"
    elif month in [6, 7, 8, 9]:
        return "Monsoon"
    else:
        return "Post-Monsoon"

time_features["season"] = (
    time_features["month"]
    .apply(get_indian_season)
)
```

```python
time_features["month_sin"] = np.sin(
    2 * np.pi * time_features["month"] / 12
)

time_features["month_cos"] = np.cos(
    2 * np.pi * time_features["month"] / 12
)
```

```python
category_avg_demand = (
    time_features
    .groupby("Product Category")["units_sold"]
    .transform("mean")
)

time_features["seasonality_index"] = (
    time_features["units_sold"] /
    category_avg_demand
) * 100
```

```python
time_features["month_effect"] = (
    time_features["units_sold"] /
    category_avg_demand
) * 100
```

```python
quarter_avg = (
    time_features
    .groupby(["Product Category", "quarter"])["units_sold"]
    .transform("mean")
)

time_features["quarter_effect"] = (
    quarter_avg /
    category_avg_demand
) * 100
```

```python
event_relevance = {
    "Diwali": {
        "Electronics": "High",
        "Clothing": "High",
        "Beauty": "High"
    },
    "Holi": {
        "Electronics": "Low",
        "Clothing": "Medium",
        "Beauty": "Medium"
    },
    "Eid al-Fitr": {
        "Electronics": "Medium",
        "Clothing": "High",
        "Beauty": "High"
    },
    "Raksha Bandhan": {
        "Electronics": "Medium",
        "Clothing": "High",
        "Beauty": "Medium"
    },
    "Dussehra": {
        "Electronics": "High",
        "Clothing": "Medium",
        "Beauty": "Medium"
    },
    "Christmas": {
        "Electronics": "Medium",
        "Clothing": "Medium",
        "Beauty": "Medium"
    },
    "Independence Day": {
        "Electronics": "Low",
        "Clothing": "Low",
        "Beauty": "Low"
    },
    "Republic Day": {
        "Electronics": "Low",
        "Clothing": "Low",
        "Beauty": "Low"
    }
}
```

```python
def get_month_event_features(year_month, category):

    month_start = pd.Timestamp(year_month + "-01")
    month_end = month_start + pd.offsets.MonthEnd(1)

    relevant = []

    for _, event in events.iterrows():

        event_date = event["event_start"]

        relevance = event_relevance.get(
            event["event_name"], {}
        ).get(category, "Low")

        relevance_score = {
            "Low": 25,
            "Medium": 50,
            "High": 100
        }[relevance]

        # Event occurs inside this month
        if month_start <= event_date <= month_end:

            relevant.append({
                "event_name": event["event_name"],
                "event_type": event["event_type"],
                "relevance": relevance,
                "relevance_score": relevance_score,
                "days_to_event": 0,
                "days_after_event": 0,
                "period": "Event",
                "tentative": event["date_tentative"]
            })

        # Month is within 30 days BEFORE event
        elif 0 < (event_date - month_end).days <= 30:

            relevant.append({
                "event_name": event["event_name"],
                "event_type": event["event_type"],
                "relevance": relevance,
                "relevance_score": relevance_score,
                "days_to_event": (event_date - month_end).days,
                "days_after_event": 0,
                "period": "Pre-Event",
                "tentative": event["date_tentative"]
            })

        # Month is within 15 days AFTER event
        elif 0 < (month_start - event_date).days <= 15:

            relevant.append({
                "event_name": event["event_name"],
                "event_type": event["event_type"],
                "relevance": relevance,
                "relevance_score": relevance_score,
                "days_to_event": 0,
                "days_after_event": (month_start - event_date).days,
                "period": "Post-Event",
                "tentative": event["date_tentative"]
            })

    # No relevant event
    if not relevant:

        return {
            "event_active": 0,
            "event_count": 0,
            "event_relevance_score": 0,
            "days_to_event": np.nan,
            "days_after_event": np.nan,
            "event_period": "None",
            "event_names": "None",
            "event_tentative": 0,
            "pre_event": 0,
            "post_event": 0
        }

    # Highest relevance event
    period_priority = {
        "Event": 3,
        "Pre-Event": 2,
        "Post-Event": 1
    }

    best = max(
        relevant,
        key=lambda x: (
            period_priority[x["period"]],
            x["relevance_score"]
        )
    )

    return {
        "event_active": int(
              any(x["period"] == "Event" for x in relevant)
          ),

        "event_count": len(relevant),

        "event_relevance_score": max(
            x["relevance_score"]
            for x in relevant
        ),

        "days_to_event": min(
            [
                x["days_to_event"]
                for x in relevant
                if x["days_to_event"] > 0
            ],
            default=0
        ),

        "days_after_event": min(
            [
                x["days_after_event"]
                for x in relevant
                if x["days_after_event"] > 0
            ],
            default=0
        ),

        "event_period": best["period"],

        "event_names": ", ".join(
            sorted(
                set(x["event_name"] for x in relevant)
            )
        ),

        "event_tentative": int(
            any(x["tentative"] for x in relevant)
        ),

        "pre_event": int(
            any(x["period"] == "Pre-Event" for x in relevant)
        ),

        "post_event": int(
            any(x["period"] == "Post-Event" for x in relevant)
        )
    }
```

```python
import requests
import pandas as pd
import time

session = requests.Session()

def fetch_world_bank_indicator(indicator_code, max_retries=4):
    url = (
        f"https://api.worldbank.org/v2/country/IND/"
        f"indicator/{indicator_code}"
    )

    params = {
        "format": "json",
        "date": "2000:2025",
        "per_page": 100
    }

    for attempt in range(max_retries):
        try:
            response = session.get(
                url,
                params=params,
                timeout=(15, 90)  # 15 sec connection, 90 sec read
            )
            response.raise_for_status()

            result = response.json()

            if not isinstance(result, list) or len(result) < 2:
                raise ValueError("Unexpected World Bank API response")

            rows = result[1] or []

            return pd.DataFrame([
                {
                    "year": int(row["date"]),
                    "value": row["value"]
                }
                for row in rows
            ])

        except (
            requests.exceptions.Timeout,
            requests.exceptions.ConnectionError
        ) as error:
            print(
                f"{indicator_code}: attempt "
                f"{attempt + 1}/{max_retries} failed: {error}"
            )

            if attempt < max_retries - 1:
                time.sleep(2 ** attempt)

    raise RuntimeError(
        f"Could not download {indicator_code} after "
        f"{max_retries} attempts. Check Colab internet access."
    )
```

```python
indicators = {
    "total_population": "SP.POP.TOTL",
    "population_growth_rate": "SP.POP.GROW",
    "urban_population_share": "SP.URB.TOTL.IN.ZS",
    "age_0_14_share": "SP.POP.0014.TO.ZS",
    "age_15_64_share": "SP.POP.1564.TO.ZS",
    "age_65_plus_share": "SP.POP.65UP.TO.ZS"
}

demographic_data = None

for feature, indicator_code in indicators.items():
    print(f"Downloading {feature}...")

    temp = fetch_world_bank_indicator(indicator_code)
    temp = temp.rename(columns={"value": feature})

    if demographic_data is None:
        demographic_data = temp
    else:
        demographic_data = demographic_data.merge(
            temp,
            on="year",
            how="outer"
        )

    time.sleep(1)

demographic_data = demographic_data.sort_values(
    "year"
).reset_index(drop=True)

print("Download complete!")
print("Shape:", demographic_data.shape)
display(demographic_data.tail(10))
```

```python
import requests
import pandas as pd
import time
import os

session = requests.Session()

def fetch_economic_indicator(indicator_code, max_retries=4):
    url = (
        f"https://api.worldbank.org/v2/country/IND/"
        f"indicator/{indicator_code}"
    )

    for attempt in range(max_retries):
        try:
            response = session.get(
                url,
                params={
                    "format": "json",
                    "date": "2000:2025",
                    "per_page": 100
                },
                timeout=(15, 90)
            )

            response.raise_for_status()
            result = response.json()

            if not isinstance(result, list) or len(result) < 2:
                raise ValueError("Unexpected API response")

            rows = result[1] or []

            return pd.DataFrame([
                {
                    "year": int(row["date"]),
                    "value": row["value"]
                }
                for row in rows
            ])

        except (
            requests.exceptions.Timeout,
            requests.exceptions.ConnectionError
        ) as error:
            print(
                f"{indicator_code}: attempt "
                f"{attempt + 1}/{max_retries} failed"
            )

            if attempt < max_retries - 1:
                time.sleep(2 ** attempt)

    raise RuntimeError(
        f"Unable to download {indicator_code}. "
        "Check the connection before retrying."
    )
```

```python
def calculate_marketing_features(
    marketing_budget,
    advertising_budget,
    campaign_reach,
    distribution_coverage,
    ecommerce_availability,
    retail_availability,
    seller_count,
    delivery_time_days,
    expected_stock_availability
):
    marketing_intensity = (
        advertising_budget / marketing_budget
        if marketing_budget and marketing_budget > 0
        else None
    )

    availability_index = (
        (
            distribution_coverage
            + ecommerce_availability * 100
            + retail_availability * 100
            + expected_stock_availability
        ) / 4
        if all(
            x is not None
            for x in [
                distribution_coverage,
                ecommerce_availability,
                retail_availability,
                expected_stock_availability
            ]
        )
        else None
    )

    return {
        "marketing_intensity": marketing_intensity,
        "availability_index": availability_index
    }
```

```python

import pandas as pd
from pathlib import Path

base = Path("/content/drive/MyDrive/Pre_Product_V1")
clean = base / "02_Cleaned_Data"
feat = base / "03_Feature_Engineered"
out = base / "04_Master_Dataset"

out.mkdir(parents=True, exist_ok=True)

# 1. Monthly demand = base table
demand = pd.read_parquet(feat / "Demand" / "monthly_demand.parquet")
demand["year_month"] = demand["year_month"].astype(str)
demand["category_key"] = demand["Product Category"].astype(str).str.strip()

# 2. Trends: standardize category/month join keys
trends = pd.read_parquet(feat / "Interest_Trends" / "consumer_interest_trends.parquet")
trends["year_month"] = trends["year_month"].astype(str)
trends["category_key"] = trends["category"].astype(str).str.strip()
trends = trends.drop(columns=["category"], errors="ignore")

# 3. Seasonality monthly features
season = pd.read_parquet(
    feat / "Time_Seasonality" / "time_seasonality_features.parquet"
)
season["year_month"] = season["year_month"].astype(str)
season["category_key"] = season["Product Category"].astype(str).str.strip()
season = season.drop(columns=["Product Category"], errors="ignore")

# Keep only one copy of columns already present in demand.
season_extra = [
    c for c in season.columns
    if c not in demand.columns or c in ["year_month", "category_key"]
]
season = season[season_extra]

# 4. Events: category-month level
events = pd.read_parquet(feat / "Events" / "future_event_features.parquet")
events["year_month"] = events["year_month"].astype(str)
events["category_key"] = events["category"].astype(str).str.strip()
events = events.drop(columns=["category"], errors="ignore")

# Ensure join keys are unique before merging.
def check_unique_keys(df, keys, name):
    duplicated = df.duplicated(keys).sum()
    print(f"{name}: rows={len(df)}, duplicate join keys={duplicated}")
    if duplicated:
        raise ValueError(
            f"{name} has duplicate join keys {keys}; fix before merging."
        )

check_unique_keys(demand, ["year_month", "category_key"], "Demand")
check_unique_keys(trends, ["year_month", "category_key"], "Trends")
check_unique_keys(season, ["year_month", "category_key"], "Seasonality")
check_unique_keys(events, ["year_month", "category_key"], "Events")

# 5. Merge only category-month data
master = demand.copy()
master = master.merge(
    trends, on=["year_month", "category_key"],
    how="left", validate="one_to_one", suffixes=("", "_trend")
)
master = master.merge(
    season, on=["year_month", "category_key"],
    how="left", validate="one_to_one", suffixes=("", "_season")
)
master = master.merge(
    events, on=["year_month", "category_key"],
    how="left", validate="one_to_one", suffixes=("", "_event")
)

# 6. Add year-level demographics and economy
master["year"] = master["year_month"].str[:4].astype(int)

demo = pd.read_parquet(
    feat / "Demographics" / "demographic_features_prototype.parquet"
)
economy = pd.read_parquet(
    feat / "Economy" / "economic_features_prototype.parquet"
)

demo["year"] = pd.to_numeric(demo["year"], errors="coerce")
economy["year"] = pd.to_numeric(economy["year"], errors="coerce")

master = master.merge(
    demo, on="year", how="left",
    validate="many_to_one", suffixes=("", "_demo")
)
master = master.merge(
    economy, on="year", how="left",
    validate="many_to_one", suffixes=("", "_economy")
)

# 7. Save category-month master dataset
master_path = out / "category_month_master_v1.parquet"
master.to_parquet(master_path, index=False)

# Save CSV preview too
master.head(100).to_csv(out / "category_month_master_v1_preview.csv", index=False)

print("\nMASTER DATASET CREATED")
print("Rows:", len(master))
print("Columns:", len(master.columns))
print("Saved:", master_path)
print("Preview:", out / "category_month_master_v1_preview.csv")
display(master.head())
```

```python
import pandas as pd
from pathlib import Path

BASE = Path("/content/drive/MyDrive/Pre_Product_V1")
FE = BASE / "03_Feature_Engineered"

demand = pd.read_parquet(FE / "Demand/monthly_demand.parquet")
interest = pd.read_parquet(
    FE / "Interest_Trends/interest_demand_alignment.parquet"
)
season = pd.read_parquet(
    FE / "Time_Seasonality/time_seasonality_features.parquet"
)
events = pd.read_parquet(
    FE / "Events/future_event_features.parquet"
)

# 1. Check category-month key uniqueness
checks = [
    ("Demand", demand, ["year_month", "Product Category"]),
    ("Interest", interest, ["year_month", "source_category"]),
    ("Seasonality", season, ["year_month", "Product Category"]),
    ("Events", events, ["year_month", "category"])
]

for name, df, keys in checks:
    print(f"\n{name}")
    print("Rows:", len(df))
    print("Duplicate keys:", df.duplicated(keys).sum())
    print("Month range:", df["year_month"].min(),
          "to", df["year_month"].max())
    print("Unique months:", df["year_month"].nunique())

# 2. Map demand categories to interest categories
mapping = {
    "Beauty": "Personal Care",
    "Clothing": "Clothing & Fashion",
    "Electronics": "Electronics"
}

demand_keys = set(
    zip(demand["year_month"], demand["Product Category"])
)

interest_keys = set(
    zip(interest["year_month"], interest["source_category"])
)

season_keys = set(
    zip(season["year_month"], season["Product Category"])
)

event_keys = set(
    zip(events["year_month"], events["category"])
)

print("\n--- KEY COVERAGE ---")
print("Demand keys:", len(demand_keys))
print("Interest matching keys:", len(demand_keys & interest_keys))
print("Seasonality matching keys:", len(demand_keys & season_keys))

mapped_event_keys = {
    (month, category)
    for month, category in event_keys
    if category in mapping
}

print("Event matching keys:", len(demand_keys & mapped_event_keys))

print("\n--- INTEREST CATEGORY MAPPING ---")
for source, target in mapping.items():
    actual = set(
        interest.loc[
            interest["source_category"] == source,
            "category"
        ].dropna().unique()
    )
    print(f"{source} -> expected {target}; found: {actual}")
```

```python
import pandas as pd
from pathlib import Path

BASE = Path("/content/drive/MyDrive/Pre_Product_V1")
FE = BASE / "03_Feature_Engineered"
OUT = BASE / "04_Master_Dataset"

OUT.mkdir(parents=True, exist_ok=True)

# Load only the three validated datasets
demand = pd.read_parquet(
    FE / "Demand/monthly_demand.parquet"
)

interest = pd.read_parquet(
    FE / "Interest_Trends/interest_demand_alignment.parquet"
)

season = pd.read_parquet(
    FE / "Time_Seasonality/time_seasonality_features.parquet"
)

# 1. Keep only interest-specific features
interest_features = interest[
    [
        "year_month",
        "source_category",
        "search_interest",
        "interest_growth",
        "trend_momentum",
        "trend_index",
        "demand_index",
        "interest_demand_gap"
    ]
].copy()

# Rename source category to match demand category
interest_features = interest_features.rename(
    columns={"source_category": "Product Category"}
)

# 2. Keep only seasonality-specific features
season_features = season[
    [
        "year_month",
        "Product Category",
        "date",
        "year",
        "month",
        "quarter",
        "season",
        "month_sin",
        "month_cos",
        "seasonality_index",
        "month_effect",
        "quarter_effect"
    ]
].copy()

# 3. Verify unique keys before merging
keys = ["year_month", "Product Category"]

assert not demand.duplicated(keys).any(), \
    "Duplicate keys found in demand"

assert not interest_features.duplicated(keys).any(), \
    "Duplicate keys found in interest"

assert not season_features.duplicated(keys).any(), \
    "Duplicate keys found in seasonality"

# 4. Merge with one-to-one validation
master = demand.merge(
    interest_features,
    on=keys,
    how="left",
    validate="one_to_one",
    indicator="_interest_match"
)

assert (master["_interest_match"] == "both").all(), \
    "Some demand rows did not match interest data"

master = master.drop(columns="_interest_match")

master = master.merge(
    season_features,
    on=keys,
    how="left",
    validate="one_to_one",
    indicator="_season_match"
)

assert (master["_season_match"] == "both").all(), \
    "Some demand rows did not match seasonality data"

master = master.drop(columns="_season_match")

# 5. Final checks
assert len(master) == len(demand), \
    "Unexpected row count after merge"

assert not master.duplicated(keys).any(), \
    "Duplicate category-month keys in master"

# 6. Save master dataset
parquet_path = OUT / "category_month_master_v1.parquet"
csv_path = OUT / "category_month_master_v1.csv"

master.to_parquet(parquet_path, index=False)
master.to_csv(csv_path, index=False)

print("MASTER DATASET CREATED SUCCESSFULLY")
print("Shape:", master.shape)
print("Rows:", len(master))
print("Columns:", len(master.columns))
print("Duplicate keys:", master.duplicated(keys).sum())

print("\nMissing values:")
print(master.isnull().sum()[master.isnull().sum() > 0])

print("\nSaved files:")
print(parquet_path)
print(csv_path)

print("\nPreview:")
display(master.head())
```

```python
import pandas as pd
from pathlib import Path

BASE = Path("/content/drive/MyDrive/Pre_Product_V1")
path = BASE / "04_Master_Dataset/category_month_master_v1.parquet"

df = pd.read_parquet(path)

# 1. Basic information
print("SHAPE:", df.shape)
print("\nCOLUMNS:")
print(df.columns.tolist())

# 2. Missing values
print("\nMISSING VALUES:")
print(df.isnull().sum()[df.isnull().sum() > 0])

# 3. Category and month coverage
print("\nROWS PER CATEGORY:")
print(df.groupby("Product Category").size())

print("\nMONTH RANGE:")
print(df["year_month"].min(), "to", df["year_month"].max())

print("\nDUPLICATE KEYS:")
print(df.duplicated(
    ["year_month", "Product Category"]
).sum())

# 4. Check chronological ordering
df["_month"] = pd.to_datetime(df["year_month"])

print("\nCHRONOLOGICALLY SORTED:")
print(
    df["_month"].is_monotonic_increasing
    if df["Product Category"].nunique() == 1
    else df.sort_values(
        ["Product Category", "_month"]
    ).groupby("Product Category")["_month"]
     .apply(lambda x: x.is_monotonic_increasing).to_dict()
)

# 5. Inspect demand target candidates
print("\nDEMAND SUMMARY:")
print(df[[
    "units_sold",
    "total_sales",
    "demand_growth",
    "demand_moving_avg_3m",
    "category_demand_index"
]].describe().round(2))

# 6. Save audit report
audit_path = BASE / "04_Master_Dataset/master_dataset_audit.txt"

with open(audit_path, "w") as f:
    f.write(f"Shape: {df.shape}\n\n")
    f.write("Columns:\n" + "\n".join(df.columns) + "\n\n")
    f.write("Missing values:\n")
    f.write(df.isnull().sum().to_string())
    f.write("\n\nDuplicate keys: ")
    f.write(str(df.duplicated(
        ["year_month", "Product Category"]
    ).sum()))

print("\nAudit report saved:", audit_path)
```
